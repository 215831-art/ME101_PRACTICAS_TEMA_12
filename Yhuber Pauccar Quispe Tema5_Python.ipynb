```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("estudiantes_limpio.csv")

# Estadística descriptiva de la columna nota
media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std()
var_poblacional = np.var(df["nota"])
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1

# Imprimir resultados
print("Media:", media)
print("Mediana:", mediana)
print("Desviación estándar muestral:", std_muestral)
print("Varianza poblacional:", var_poblacional)
print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

```python
# Asimetría y curtosis con pandas
skew_pandas = df["nota"].skew()
kurt_pandas = df["nota"].kurt()

# Asimetría y curtosis con scipy
skew_scipy = stats.skew(df["nota"])
kurt_scipy = stats.kurtosis(df["nota"])

# Imprimir resultados
print("Asimetría (pandas):", skew_pandas)
print("Curtosis (pandas):", kurt_pandas)
print("Asimetría (scipy):", skew_scipy)
print("Curtosis (scipy):", kurt_scipy)
```

```python
def resumen_estadistico(serie, decimales=4):
    n = int(serie.count())
    media = float(serie.mean())
    mediana = float(serie.median())
    desv_std = float(serie.std())
    cv_pct = (desv_std / media) * 100

    return {
        "n": n,
        "media": round(media, decimales),
        "mediana": round(mediana, decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round(cv_pct, decimales)
    }

print(resumen_estadistico(df["nota"]))
```

```python
# Definir función para clasificar la dispersión

def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"


# Bucle for sobre las dos columnas

for columna in ["nota", "asistencia_pct"]:
    resumen = resumen_estadistico(df[columna])
    dispersion = clasificar_dispersion(resumen["cv_pct"])
    
    print("Columna:", columna)
    print("Resumen:", resumen)
    print("Dispersión:", dispersion)
    print()
```
