```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread("Pin on Lionel messi.jpg")

if imagen is None:
    print("No se encontró la imagen. Revisa el nombre del archivo.")
else:
    imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
    plt.imshow(imagen_rgb)
    plt.title("Imagen original de Messi")
    plt.axis("off")
    plt.show()
```

```python
alto, ancho = imagen_rgb.shape[:2]

ampliada = cv2.resize(
    imagen_rgb, (ancho*2, alto*2),
    interpolation=cv2.INTER_CUBIC
)

reducida = cv2.resize(
    imagen_rgb, (ancho//2, alto//2),
    interpolation=cv2.INTER_AREA
)

fig, ax = plt.subplots(1, 3, figsize=(12, 5))
ax[0].imshow(imagen_rgb)
ax[0].set_title("Original")
ax[1].imshow(ampliada)
ax[1].set_title("Ampliada")
ax[2].imshow(reducida)
ax[2].set_title("Reducida")

for a in ax:
    a.axis("off")
plt.show()
```

```python
alto, ancho = imagen_rgb.shape[:2]
centro = (ancho//2, alto//2)

matriz = cv2.getRotationMatrix2D(centro, 30, 1)
rotada = cv2.warpAffine(imagen_rgb, matriz, (ancho, alto))

plt.figure(figsize=(6, 8))
plt.imshow(rotada)
plt.title("Imagen de Messi rotada 30 grados")
plt.axis("off")
plt.show()
```

```python
alto, ancho = imagen_rgb.shape[:2]

matriz = np.float32([[1, 0, 100], [0, 1, 70]])
movida = cv2.warpAffine(imagen_rgb, matriz, (ancho, alto))

plt.imshow(movida)
plt.title("Imagen desplazada")
plt.axis("off")
plt.show()
```

```python
alto, ancho = imagen_rgb.shape[:2]

matriz = np.float32([
    [1, 0.2, 0],
    [0, 1, 0]
])

deformada = cv2.warpAffine(imagen_rgb, matriz, (ancho, alto))

plt.imshow(deformada)
plt.title("Imagen deformada")
plt.axis("off")
plt.show()
```

```python
normalizada = cv2.normalize(
    imagen_rgb.astype("float32"),
    None, 0, 1, cv2.NORM_MINMAX
)

plt.imshow(normalizada)
plt.title("Imagen normalizada")
plt.axis("off")
plt.show()
```

```python
gris = cv2.cvtColor(imagen_rgb, cv2.COLOR_RGB2GRAY)
bordes = cv2.Canny(gris, 100, 200)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(imagen_rgb)
ax[0].set_title("Imagen original")
ax[1].imshow(bordes, cmap="gray")
ax[1].set_title("Bordes de la imagen")

for a in ax:
    a.axis("off")
plt.show()
```

```python
desenfocada = cv2.GaussianBlur(imagen_rgb, (9, 9), 0)

fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(imagen_rgb)
ax[0].set_title("Original")
ax[1].imshow(desenfocada)
ax[1].set_title("Desenfocada")

for a in ax:
    a.axis("off")
plt.show()
```

```python
gris = cv2.cvtColor(imagen_rgb, cv2.COLOR_RGB2GRAY)
kernel = np.ones((3, 3), np.uint8)

dilatada = cv2.dilate(gris, kernel, iterations=2)
erosionada = cv2.erode(gris, kernel, iterations=2)
apertura = cv2.morphologyEx(gris, cv2.MORPH_OPEN, kernel)
cierre = cv2.morphologyEx(gris, cv2.MORPH_CLOSE, kernel)

fig, ax = plt.subplots(2, 2, figsize=(9, 9))

ax[0, 0].imshow(dilatada, cmap="gray")
ax[0, 0].set_title("Dilatación")
ax[0, 1].imshow(erosionada, cmap="gray")
ax[0, 1].set_title("Erosión")
ax[1, 0].imshow(apertura, cmap="gray")
ax[1, 0].set_title("Apertura")
ax[1, 1].imshow(cierre, cmap="gray")
ax[1, 1].set_title("Cierre")

for a in ax.flatten():
    a.axis("off")

plt.tight_layout()
plt.show()
```
