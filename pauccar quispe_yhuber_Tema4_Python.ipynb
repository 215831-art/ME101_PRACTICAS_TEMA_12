```python
import pandas as pd
import numpy as np

# Crear el diccionario con 6 estudiantes y valores faltantes
datos = {
    "nombre": ["Ana", "Luis", "María", "Carlos", "Sofía", "Pedro"],
    "curso": ["Matemática", "Estadística", "Matemática", "Programación", "Estadística", "Programación"],
    "nota": [15, 18, np.nan, 14, 17, np.nan],
    "asistencia_pct": [90, np.nan, 85, 95, 88, 92]
}

# Crear el DataFrame
df_original = pd.DataFrame(datos)

# Guardar como archivo CSV
df_original.to_csv("estudiantes.csv", index=False)

# Mostrar el DataFrame
df_original
```

```python
# Lee el archivo CSV
df = pd.read_csv("estudiantes.csv")

# Muestra la forma del DataFrame
print("Forma del DataFrame:", df.shape)

# Muestra la cantidad de valores faltantes por columna
print("\nValores faltantes por columna:")
print(df.isna().sum())
```

```python
# Elimina las filas con nota faltante
df_limpio = df.dropna(subset=["nota"])

# Rellena asistencia_pct faltante con el promedio
promedio_asistencia = df_limpio["asistencia_pct"].mean()
df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(promedio_asistencia)

# Verifica que no queden valores faltantes
print(df_limpio.isna().sum())

# Muestra el DataFrame limpio
df_limpio
```

```python
# Promedio de nota agrupado por curso
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()

print(promedio_por_curso)
```

```python
df_limpio.to_csv("estudiantes_limpio.csv", index=False)
```
