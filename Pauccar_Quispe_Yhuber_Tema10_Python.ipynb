```python

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Dataset del Tema 9: monitoreo ambiental
datos = {
    "estacion": ["Centro", "San Jerónimo", "Wanchaq",
                 "San Sebastián", "Poroy"],
    "pm10_ugm3": [45, 28, 62, 35, np.nan],
    "ruido_db": [72, 58, 81, 65, 52],
    "tipo_zona": ["Urbana", "Valle", "Urbana",
                  "Urbana", "Altoandina"]
}

# Crear DataFrame
df = pd.DataFrame(datos)

# Seleccionar variables numéricas válidas
variables = df[["pm10_ugm3", "ruido_db"]]

# Calcular matriz de correlación
corr = variables.corr()

# Mostrar resultados
print("DATOS DEL TEMA 9")
display(df)

print("MATRIZ DE CORRELACIÓN")
print(corr.round(4))
```

```python

# Crear figura
fig, ax = plt.subplots(figsize=(8, 6))

# Construir heatmap anotado
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    cbar_kws={"label": "Correlación de Pearson"},
    ax=ax
)

# Agregar título
plt.title("Heatmap de correlación - Datos ambientales", fontsize=14)

# Ajustar presentación
plt.tight_layout()

# Guardar imagen
plt.savefig("heatmap_seaborn.png", dpi=300, bbox_inches="tight")

# Mostrar gráfico
plt.show()

```

```python

# Crear máscara triangular superior
mascara = np.triu(np.ones_like(corr, dtype=bool))

# Imprimir cantidad de celdas visibles
print("Celdas visibles:", (~mascara).sum(), "de", mascara.size)

# Crear figura
fig, ax = plt.subplots(figsize=(8, 6))

# Construir heatmap con máscara
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    mask=mascara,
    cbar_kws={"label": "Correlación de Pearson"},
    ax=ax
)

# Agregar título
plt.title("Heatmap ambiental con máscara triangular", fontsize=14)

# Ajustar presentación
plt.tight_layout()

# Guardar imagen
plt.savefig("heatmap_seaborn_mascara.png", dpi=300, bbox_inches="tight")

# Mostrar gráfico
plt.show()
```
